```python
import torch
from diffusers import DiffusionPipeline, DPMSolverMultistepScheduler
from diffusers.utils import export_to_video
import imageio
```

```python
# Set up the pipeline
pipe = DiffusionPipeline.from_pretrained("damo-vilab/text-to-video-ms-1.7b", torch_dtype=torch.float16, variant="fp16")
pipe.scheduler = DPMSolverMultistepScheduler.from_config(pipe.scheduler.config)

# Move to GPU if available (recommended for speed)
pipe = pipe.to("cuda" if torch.cuda.is_available() else "cpu")
```

```python

# Enter your text prompt here
prompt = "A lakeshore view of Downtown Toronto Canada with city skyline reflected in Lake Ontario"

# Generate video frames
# num_frames affects the length of your video (8 frames ≈ 1 second at default FPS)
# Lower num_inference_steps (20-30) for faster generation with slightly lower quality
# Increase height/width for better quality but at the cost of more VRAM
video_frames = pipe(
    prompt,
    num_inference_steps=30,        # Default is 50, lower for faster generation
    num_frames=80,                 # 80 frames for ~10 seconds at default FPS
    height=256,                    # Default is 320, lower for less VRAM usage
    width=256,                     # Default is 576, lower for less VRAM usage
    guidance_scale=7.5,            # How closely to follow the prompt
).frames
```

```python
import torch
import os
import numpy as np
from diffusers import DiffusionPipeline, DPMSolverMultistepScheduler
from diffusers.utils import export_to_video
from IPython.display import HTML
from base64 import b64encode
import matplotlib.pyplot as plt
```

```python
# Set device
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")
print(f"CUDA available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"CUDA device: {torch.cuda.get_device_name()}")
    print(f"CUDA memory allocated: {torch.cuda.memory_allocated() / 1024**2:.2f} MB")
    print(f"CUDA memory reserved: {torch.cuda.memory_reserved() / 1024**2:.2f} MB")

# Create output directories
os.makedirs("toronto_frames", exist_ok=True)
```

```python
# Clear CUDA cache for maximum available memory
torch.cuda.empty_cache()
print(f"After clearing cache - CUDA memory allocated: {torch.cuda.memory_allocated() / 1024**2:.2f} MB")
print(f"After clearing cache - CUDA memory reserved: {torch.cuda.memory_reserved() / 1024**2:.2f} MB")

# Load the text-to-video model (ModelScope is better than ZeroScope for quality)
print("Loading ModelScope text-to-video-ms model...")
pipe = DiffusionPipeline.from_pretrained(
    "damo-vilab/text-to-video-ms-1.7b",
    torch_dtype=torch.float16,
    variant="fp16"
)
pipe.scheduler = DPMSolverMultistepScheduler.from_config(pipe.scheduler.config)
pipe = pipe.to(device)
```

```python
# Enable memory optimizations for T4
pipe.enable_model_cpu_offload()
pipe.enable_vae_slicing()
```

```python
# Set the prompt
prompt = "A cinematic aerial view of Toronto skyline with CN Tower, beautiful sunny day, Lake Ontario in foreground, 4K, detailed"
negative_prompt = "low quality, blurry, distorted"
```

```python
# We'll create two sequences and join them for better control, but shorter
# First sequence: Toronto skyline
print("Generating first sequence (Toronto skyline)...")
skyline_frames = pipe(
    prompt="A cinematic aerial view of Toronto skyline with CN Tower, beautiful sunny day, Lake Ontario in foreground, 4K, detailed",
    negative_prompt="low quality, blurry, distorted, canadian flag",
    num_inference_steps=30,
    num_frames=15,              # First part (about 3 seconds)
    height=320,
    width=576,
    guidance_scale=7.5,
).frames
```

```python
# Second sequence: Toronto skyline with Canadian flag
print("Generating second sequence (with Canadian flag)...")
flag_frames = pipe(
    prompt="A cinematic aerial view of Toronto skyline with CN Tower, Canadian flag waving in the foreground, beautiful sunny day, Lake Ontario visible, 4K, detailed",
    negative_prompt="low quality, blurry, distorted",
    num_inference_steps=30,
    num_frames=10,              # Last 2 seconds
    height=320,
    width=576,
    guidance_scale=8.0,         # Slightly higher guidance to emphasize the flag
).frames
video_frames = skyline_frames
```

```python
# Save frames individually first
print("Saving individual frames...")
frame_paths = []

# Create folders for both sequences
os.makedirs("toronto_frames_skyline", exist_ok=True)
os.makedirs("toronto_frames_flag", exist_ok=True)
os.makedirs("toronto_frames_combined", exist_ok=True)

# Process and save skyline frames
if hasattr(skyline_frames, 'shape') and len(skyline_frames.shape) == 5:
    num_frames = skyline_frames.shape[1]
    print(f"Processing {num_frames} skyline frames...")
    for i in range(num_frames):
        frame = skyline_frames[0, i]  # Extract from 5D array
        plt.figure(figsize=(10, 6))
        plt.imshow(frame)
        plt.axis('off')
        frame_path = f"toronto_frames_skyline/frame_{i:04d}.png"
        plt.savefig(frame_path, bbox_inches='tight', pad_inches=0)
        plt.close()
        print(f"Saved skyline frame {i+1}/{num_frames}")

        # Also save to the combined folder
        combined_path = f"toronto_frames_combined/frame_{i:04d}.png"
        plt.figure(figsize=(10, 6))
        plt.imshow(frame)
        plt.axis('off')
        plt.savefig(combined_path, bbox_inches='tight', pad_inches=0)
        plt.close()
else:
    print("Unexpected skyline_frames structure")

# Process and save flag frames
if hasattr(flag_frames, 'shape') and len(flag_frames.shape) == 5:
    num_frames = flag_frames.shape[1]
    print(f"Processing {num_frames} flag frames...")
    for i in range(num_frames):
        frame = flag_frames[0, i]  # Extract from 5D array
        plt.figure(figsize=(10, 6))
        plt.imshow(frame)
        plt.axis('off')
        frame_path = f"toronto_frames_flag/frame_{i:04d}.png"
        plt.savefig(frame_path, bbox_inches='tight', pad_inches=0)
        plt.close()
        print(f"Saved flag frame {i+1}/{num_frames}")

        # Also save to the combined folder with offset
        skyline_count = skyline_frames.shape[1] if hasattr(skyline_frames, 'shape') else 28
        combined_path = f"toronto_frames_combined/frame_{skyline_count + i:04d}.png"
        plt.figure(figsize=(10, 6))
        plt.imshow(frame)
        plt.axis('off')
        plt.savefig(combined_path, bbox_inches='tight', pad_inches=0)
        plt.close()
else:
    print("Unexpected flag_frames structure")

# Get all frames from the combined folder
combined_frames = sorted([f for f in os.listdir("toronto_frames_combined") if f.endswith(".png")])
frame_paths = [os.path.join("toronto_frames_combined", f) for f in combined_frames]
print(f"Total combined frames: {len(frame_paths)}")



# Create video using ffmpeg with combined frames
print("Creating video with ffmpeg...")
output_video = "toronto_video.mp4"
# Use ffmpeg with a framerate of 5 fps for a longer video
!ffmpeg -y -framerate 5 -pattern_type glob -i "toronto_frames_combined/frame_*.png" -c:v libx264 -vf "scale=trunc(iw/2)*2:trunc(ih/2)*2" -pix_fmt yuv420p {output_video}

# Display the video
if os.path.exists(output_video):
    print("Video created successfully!")
    mp4 = open(output_video, 'rb').read()
    data_url = "data:video/mp4;base64," + b64encode(mp4).decode()

    display(HTML(f"""
    <video width=576 height=320 controls>
        <source src="{data_url}" type="video/mp4">
    </video>
    """))

    # Calculate video duration
    total_frames = len([f for f in os.listdir("toronto_frames_combined") if f.endswith(".png")])
    duration_seconds = total_frames / 5  # at 5 fps
    print(f"Video creation complete! Duration: approximately {duration_seconds:.1f} seconds")
    print("You can download the video from the files panel.")
else:
    print("Failed to create video file.")

# Display the video
if os.path.exists(output_video):
    print("Video created successfully!")
    mp4 = open(output_video, 'rb').read()
    data_url = "data:video/mp4;base64," + b64encode(mp4).decode()

    display(HTML(f"""
    <video width=576 height=320 controls>
        <source src="{data_url}" type="video/mp4">
    </video>
    """))

    print("Video creation complete! You can download the video from the files panel.")
else:
    print("Failed to create video file.")
```
